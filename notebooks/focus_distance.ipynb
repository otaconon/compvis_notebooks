# Focus distance and depth of field

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use("dark_background")
```

## Focus distance as a function of the distance traveled from the focal

Let $z_0$ denote the focus distance, and $\Delta z_i = f - z_i$ denote distance traveled from the focal length, for a lens of focal length $f = 100\text{mm}$.
We have $\frac{1}{z_0} + \frac{1}{z_i} = \frac{1}{f}$, thus $z_0 = \frac{f(\Delta z_i - f)}{\Delta z_i}$.

```python
f = 100
dz_i = f - np.arange(101, 200)
z_0 = f*(dz_i-f)/dz_i
fig, ax = plt.subplots(1)
ax.plot(dz_i, z_0)
ax.set_xlabel("$\\Delta z_i$")
ax.set_ylabel("$z_0$");
```
