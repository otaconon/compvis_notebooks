```python
from pathlib import Path
import cupy as cp
import numpy as np
```

```python
mod = cp.RawModule(code=Path("../kernels/simple.cu").read_text(), name_expressions=["vec_add<float>"])
vec_add = mod.get_function("vec_add<float>")
```

```python
N = 10
d_a = cp.ones(N, dtype=cp.float32)
d_b = cp.full(N, 2.0, dtype=cp.float32)
d_c = cp.empty(N, dtype=cp.float32)

vec_add(((N + 255) // 256,), (256,), (d_a, d_b, d_c, np.int32(N)))
h_c = cp.asnumpy(d_c)
```
