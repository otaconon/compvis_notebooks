```python
from pathlib import Path
import cupy as cp
import numpy as np
from cuda_helpers.matrix import to_view
```

```python
mod = cp.RawModule(code=Path("../kernels/math.cu").read_text(), name_expressions=["add_kernel<float>"])
add_kernel = mod.get_function("add_kernel<float>")
```

```python
N = 10
d_A = cp.ones(N, dtype=cp.float32)
d_B = cp.full(N, 2.0, dtype=cp.float32)
d_C = cp.empty(N, dtype=cp.float32)

add_kernel(((N + 255) // 256,), (256,), (to_view(d_A), to_view(d_B), to_view(d_C)))
h_C = cp.asnumpy(d_C)
```
